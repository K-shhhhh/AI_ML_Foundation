Q1. For the following dataset generated using NumPy: np.random.normal(loc=70, scale=10, size=200)

- Plot a histogram of scores
- Draw a vertical line for the mean & try to interpret the results.

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.normal(loc=70, scale=10, size=200) # mean, std, total
mean_data = data.mean()
plt.hist(data,edgecolor="black")
plt.xlabel("Numpy Data set")
plt.title("Question 1 : random numpy graph")
plt.grid(True)
plt.axvline(
    x=mean_data,
    color = "red",
    label = "mean of the data",
    linewidth = 3,
    linestyle = "--"   
)
plt.legend()
plt.show()
```

## Interpretation

- The Shape: The histogram forms a "bell curve" shape, which is the signature of a normal distribution. Most of the data clusters around the middle.

- The Center (loc=70): The red dashed line (the mean) lands right at exactly 70. This represents the average score of the dataset.

- The Spread (scale=10): The data spreads out from about 40 to 100. Because the standard deviation (scale) is 10, statistical rules tell us that the vast majority (about 68%) of all the generated scores fall exactly between 60 and 80.
