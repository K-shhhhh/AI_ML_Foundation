### Q5. For the Seaborn tips dataset, do the following:
- Create a pivot table of average tip :-
    1. Rows: day
    2. Columns: time
- Plot the pivot table as a heatmap
- Annotate values
- Interpret which day and time combination performs best

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="ticks")

tips = sns.load_dataset("tips")
tips.head()

tips_pivot = tips.pivot_table( # because we have duplicate values at time and days column we use pivot table
    index="day",
    columns="time",
    values="tip",
    aggfunc="mean",# using mean values so that all the duplicates value combine into one and the data has no duplicates
    observed=False # ignore
)

print(tips_pivot)

sns.heatmap(
    data=tips_pivot,
    cmap="coolwarm",
    annot=True
)

plt.tight_layout()
plt.xlabel("Time")
plt.ylabel("Days")

# sunday dinner has the highest average tip value (3.3)
# friday lunch has the lowest average tip value (2.4)
```
