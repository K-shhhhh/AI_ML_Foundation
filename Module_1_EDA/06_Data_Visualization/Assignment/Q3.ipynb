### Q3. For the Seaborn tips dataset, do the following:
- Create a box plot of total_bill for each day
- Add hue="sex"
- Identify which day has the highest median bill
- Comment on outliers

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="ticks")

tips = sns.load_dataset("tips")
tips.head()

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.title("Box plot for total bill paid on each day by men and women")
plt.tight_layout()
plt.grid(True)
plt.legend(loc="upper left")

# sunday has highest median bill
# men have total 5 outliers, 1 on thursday, 2 on saturdy and 2 on sunday
# women have 4 outliers, 3 on thursday and 1 on saturday
# the highest outlier has the value around 51 (given by men) closely followed by 49 (twice by men) 
# there are no lower side outliers, only upper side outliers
```
