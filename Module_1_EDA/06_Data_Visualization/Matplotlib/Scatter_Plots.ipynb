### Shows the relationship between 2 variables by plotting points on a 2d plain
- used mostly in :-

  1. outlier detection
  2. co related data
  3. find clusters in the data

```python
import matplotlib.pyplot as plt

people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

# 120 bpm is like average blood pressure

colors_1 = ["red", "blue", "violet", "black", "grey", "pink", "violet", "grey", "black", "red"]
colors_2 = ["green" if bp<135 else "red" for bp in blood_pressure]

# plt.scatter(age, blood_pressure, color = "red") # normally customizing
# plt.scatter(age, blood_pressure, color = colors) # passing a list of colors for customizing each point

# Q. show red for everyone whose blood pressure is greater than or equal to 135, if its less than show green
# plt.scatter(age, blood_pressure, color = colors_2) # passing colors_2


# Q. if the value of bp is low, then the size of the point should be accordingly
# plt.scatter(age, blood_pressure, color = colors_2, s = blood_pressure) # size of data point according to the bp value

# Q. make a scatter plot where the points are a triangle rather than a circle
# plt.scatter(age, blood_pressure, color = colors_2, s = blood_pressure, marker = "<") # data points are triangles

# Q. make a scatter plot where the points a little more transparent than usual
# plt.scatter(age, blood_pressure, color = colors_2, s = blood_pressure, marker = "<", alpha = 0.5) # alpha value decides transparency

# we can go to the matplotlib.org/stable/users/explain/colors/colormaps website and look for customised colormaps for our data
plt.scatter(age, blood_pressure, s = blood_pressure, cmap = "OrRd", c = blood_pressure) # OrRd is a customised set of colors
                                                                                        # c value decides how the cmap gets distributed to our data points

plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure (in BPM)" )
plt.grid()
plt.colorbar(label="BP") # shows what each color data point indicates 

# Q. show which person denotes which exact data point for the whole scatter plot

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i],blood_pressure[i]), xytext = (age[i]-3,blood_pressure[i] + 1.5))
    # the function is like plt.annotate("text",xy = "position of our point", xytext = "position of our text related to the point")

plt.xlim(min(age), max(age) + 10)
plt.ylim(min(blood_pressure), max(blood_pressure) + 5)
```

### multiple datasets on a scatter plot

```python
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp, winter_humidity, color="blue", label = "Winter")
plt.scatter(summer_temp, summer_humidity, color="orange", label = "Summer")

plt.title("Summer Vs Winter (Humidity and Temperature) ")
plt.xlabel("Temperature in °C")
plt.ylabel("Humidity in %")
plt.legend()
plt.show()
```
