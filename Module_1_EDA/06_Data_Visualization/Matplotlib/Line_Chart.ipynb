```python
# matplotlib as a whole is a pretty big library, we use a special module called pyplot which will act as a toolkit for drawing

import matplotlib.pyplot as plt

X = [1,2,3,4]
Y = [5,6,7,8]

plt.plot(X,Y) # 1. .plot() is a function which draws a line chart, we also have to pass the coordinates

plt.show() # very useful in python scripts, read about this in the sample_lineplot.py file
```

## plotting some real data

```python
# Oscar Winner Movies

oscar_movies = ["The Dark Knight", "The Hurt Locker",  "The King's Speech", "The Artist", "Argo"]
oscar_years = [2008, 2009, 2010, 2011, 2012] # years in which these were released
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

# Q. we want to plot a graph to show how much revenue was generated in each year
# using simple line graph

plt.plot(oscar_years, oscar_revenue) # this graph is very plain and misleading, therefore we will use some additional methods

plt.title("Oscar Winning Movies Revenue Per Year") # 1. title
plt.xlabel("Years") # 2. x-axis label
plt.ylabel("Revenue (in $M)") # 3. y-axis label

# this was just plotting single dataset on line graph, next we will see multiple datasets plotting on a single line graph
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(oscar_years, oscar_revenue, label="Oscar Movies") # previous graph, but using label so that its easy to identify
plt.plot(non_oscar_years, non_oscar_revenue, label="Non-Oscar Movies") # new graph,  but using label so that its easy to identify

plt.title("Oscar Winning Movies vs Non Oscar Winning Movies Per Year Revenue")
plt.xlabel("Years") 
plt.ylabel("Revenue (in $M)") 
plt.legend() # 4. legend actually prints the labels we gave to each graph while plotting it 
             # note : we can give the location for legend but we dont cuz it automatically finds the best one  

# plt.legend(loc="upper right")
# plt.legend(loc="upper left")
# plt.legend(loc="center")
```

```python
# format strings : use to customize the actual lines on our line chart
# it has a part of color, marker(how do we want to highlight our datapoints) and line

# visit this site for documentation and information : https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html#

# customise single element
plt.plot(oscar_years, oscar_revenue, color = "green", marker = "<", linestyle = "--", linewidth = "0.5", label="Oscar Movies") 

# customise multiple elements in a single string alltogether 
plt.plot(non_oscar_years, non_oscar_revenue, "o--r", label="Non-Oscar Movies") 

plt.title("Oscar Winning Movies vs Non Oscar Winning Movies Per Year Revenue")
plt.xlabel("Years") 
plt.ylabel("Revenue (in $M)") 
plt.legend()
```

## other methods and attributes


```python
plt.plot(oscar_years, oscar_revenue, label="Oscar Movies") 
plt.plot(non_oscar_years, non_oscar_revenue, label="Non-Oscar Movies") 

plt.title("Oscar Winning Movies vs Non Oscar Winning Movies Per Year Revenue")
plt.xlabel("Years") 
plt.ylabel("Revenue (in $M)") 
plt.legend()

print(plt.style.available) # 1. shows us how many default style templates do we have
plt.style.use("seaborn-v0_8-whitegrid") # 2. this is how we actually use the style templates shown above
plt.grid(True) # 3. shows a grid in the chart so that we can identify easily the data points
plt.tight_layout() # 4. when values of x or y axis sometimes overlap, we use this to give them proper spacing
plt.savefig("line_chart_1.png") # 5. donwloads our graph in png format

plt.show()
```
