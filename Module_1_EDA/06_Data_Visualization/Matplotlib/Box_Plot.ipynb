### statistical visualization that summarizes the distribution of a dataset using 5 key numbers :-

    1. minimum = the lowest outlier
    2. first quartile (Q1) = 25th percentile                                    
    3. median (Q2) = centre vale of data after sorting, 50th percentile         
    4. third quartile (Q3) = 75th quartile                                      
    5. maximum = highest outlier

- also used for outlier detection
- used generally to check our data is skewed in which direction and till what extent
- When calculating Q1 OR Q2, sometimes median is taken in consideration and sometimes not, in matplotlib it is taken by default
- ALWAYS remember:-

    1. 25% of data <= Q1
    2. 50% of data <= Q2
    3. 75% of data <= Q3



- lowest whisker : Q1 - [(Whisker length) * (Inter Quartile Range)]
- Highest whisker : Q2 + [(Whisker length) * (Inter Quartile Range)]
- whisker length => generally is 1.5 but we can change to 2 or 2.5 as well
- inter quartile range => the vertical length of the box i.e. in our case Q3-Q1 value
- When we plot the graph, we dont plot the lowest and highest whisker, but the values in our data nearest to these


```python
import matplotlib.pyplot as plt

data = [7, 8, 5, 6, 9, 4, 10, 12, 15] # 20 is the outlier as it doesnt come between lowest and highest whisker

# median(Q2) = 8
# Q1 = 6
# Q3 = 10
# lowest whisker = 0
# highest whisker = 16
# min = 4
# max = 15


plt.boxplot(data)
plt.grid(True)
```

## Additional operations

```python
# 1. multiple datasets

import numpy as np

g1 = np.random.normal(50,10,100) # mean, std, total nums
g2 = np.random.normal(60,15,100)

plt.boxplot([g1,g2], tick_labels = ["group 1", "group 2"])

plt.grid(True)
plt.show()
```

```python
# 2. Horizontal box plots

plt.boxplot([g1,g2], tick_labels = ["group 1", "group 2"], vert=False)

plt.grid(True)
plt.show()
```

```python
# 3. show mean of the data

plt.boxplot([g1,g2], tick_labels = ["group 1", "group 2"], vert=False, showmeans=True)

plt.grid(True)
plt.show()
```

```python
# 4. change whisker length

plt.boxplot(data, whis=2.0)
plt.grid(True)
plt.show
```
