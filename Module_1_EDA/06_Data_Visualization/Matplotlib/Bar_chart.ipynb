### Bar chart displays categorical data using rectangular bars, with height/ width of each bar showing data value
- Bar charts are of 2 types :-
    1. vertical (majorly we use these)
    2. horizontal

```python
# plotting the bar chart

import matplotlib.pyplot as plt

oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

plt.bar(oscar_years, oscar_revenue, color = "purple") # function to draw a bar graph
plt.title("Revenue of Oscar Winning Movies Per Year")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")

# Q. print the exact values of the reveneus on the bars

for i in range(len(oscar_years)):
    plt.text(oscar_years[i], oscar_revenue[i] + 5, str(oscar_revenue[i]), ha = "center") # (pos on x-axis, pos on y-axis, text, align)

plt.ylim(0, max(oscar_revenue) + 100) # use to change limits on axes (lower limit, upper limit)
```

### multiple datasets on a bar chart

- we need to maniplate the width of the bar chart so that each bar is visible
- by default the value of width is 0.8, with default range being 0 to 1
- the np. array and x mathematics is important so that the 2 bars dont coincide and dont overlap

```python
# oscar movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

# non-oscar movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012] # some data points needs to be common for multiple datasets to be plotted on a single graph

import numpy as np # we need numpy to make a computational array

x = np.arange(len(years)) # represents the position of each category on x axis, that is telling python to plot bars at 0,1,2....
width = 0.4


plt.bar(x-(width/2), oscar_revenue, width, label="Oscar movies") # subtracting 0.2 so that the bar goes on left side
plt.bar(x+(width/2), non_oscar_revenue, width, label="Non-Oscar movies") # adding 0.2 so that the bar goes on the right side

plt.title("Revenue of Oscar Movies vs Non-Oscar Movies Per Year")
plt.xlabel("Years") # Use xlabel when you want to name the whole axis (e.g., "Time", "Revenue", "Years")
plt.ylabel("Revenue(in $M)")
plt.xticks(x, years) # Use xticks when you need to rename the individual data markers (e.g., swapping out 0, 1 for 2008", "2009")
plt.legend()
plt.show()
```

## Horizontal Bar Charts

- Categories on Y-Axis and Numbers/Data value on X-Axis

```python
# oscar movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.barh(oscar_movies, oscar_revenue) # using .barh signifying horizontal bar, Y-Axis values comes first here
plt.title("Revenue of Oscar Movies")
plt.xlabel("Revenue(in $M) ") 
plt.ylabel("Oscar Winning Movies")

plt.show()
```
