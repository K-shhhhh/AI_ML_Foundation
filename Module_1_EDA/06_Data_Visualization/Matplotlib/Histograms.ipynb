### shows distribution of numerical data by :-

- dividing values into bins(ranges) = x-axis
- counting how many values fall into each bin(frequency) = y-axis

### use histograms when we need to show numerical data and not categorical data

### major mistakes made while plotting a histogram:-

    1. too less bins - data has no structure, therefore cant find insights
    2. too many bins - data is very noisy i.e. outleirs are given too much importance
    

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0) # starting point is 0 so that everytime we run the data we get exact sequence of random numbers

scores = np.random.normal(70, 10, 100) # normal distribution
                                       # 70 is the mean (most values will be around 70)
                                       # 10 is the standard deviation, It dictates how far the numbers stretch away from the mean
                                       # 100 is the size, a total of 1000 numbers will be generated

# plt.hist(scores)
plt.xlabel("scores of students")
plt.ylabel("frequency of students")
plt.title("Data of marks for a class")

# customizations

# 1. colors
# plt.hist(scores, color="skyblue", edgecolor="black")

# 2. bins = range on the x axis for each bar i.e. range for continuous intervals

# plt.hist(scores, color="skyblue", edgecolor="black", bins=15)

Bins = [20,40,60,80,100] # if range is 60-80 then 60 is inclusive but 80 is exclusive but in the last range 100 will be inclusive

plt.hist(scores, color="skyblue", edgecolor="black",bins=Bins)
```

```python
# multiple datasets 

legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

# transaction amount distribution : fraud vs legit ((in $))

plt.hist(fraud_transactions, color="red", bins=20, label="Fraud Transcations",  edgecolor="black", alpha=0.5)
plt.hist(legit_transactions, color="green", bins=20, label="Legit Transactions", edgecolor="black", alpha=0.5)
plt.xlabel("Number of Transactions")
plt.ylabel("Frequency of Transactions")
plt.legend()
plt.grid(True)
plt.show()
```

### axvlines

- used to draw a vertical reference line on a plot
- common to mark cutoffs, averages, passing marks or thresholds

```python
# Q. suppose in our marks data, 33 is the passing marks, so show that in the histogram so that we can know how many students failed

marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

Bins = [20,40,60,80,100]

plt.hist(marks, color="skyblue", bins = Bins)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
plt.grid()

plt.axvline(
    x=33,
    label="passing marks (33)",
    color="red",
    linewidth="2",
    linestyle="--"
    
)
plt.legend()
plt.show()
```
