## Weekly Temperature
- compare temperature of cities over a week using subplots and loops

```python
# basically we have to create graph for each city comparing their temp on certain days

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]

fig,ax = plt.subplots(2,2) # rows and columns

# j = 1
# ax[0][0].plot(days,temperatures[j],color="green")

k = 0 # used to track the temperatures and cities list 

colors = ["blue", "red", "violet", "black"]

c = 0 # used to track the colors list

# this is the old rigid way, still very imp to know
# for i in range(2) : # rows (i) 
#     for j in range(2) : # cols (j)
#         ax[i][j].plot(days,temperatures[0],color = colors[c],marker="o")
#         ax[i][j].set_title(f"{cities[k]}")
#         # ax[i][j].set_xlabel("Days")
#         # ax[i][j].set_ylabel("Temperature (in Celcius)")
#         ax[i][j].grid(True)
#         k = k + 1 # updating it after every graph to go to the next city
#         c = c + 1 # updating it after every graph to go to the next color


# same task but in a better way 
# we flatten the 2d ax array into 1d array so that we can just use loop rather than 2
axes = ax.flatten()

for i, ax in enumerate(axes) : # enumerate function returns the index and the value at that index at the same time
                               # therefore now instead of 2*2 we will have 1 array and i will go through all of them like 1,2,3,4
    
    ax.plot(days,temperatures[i],color = colors[i],marker="o")
    ax.set_title(f"{cities[i]}")
    # ax.set_xlabel("Days")
    # ax.set_ylabel("Temperature (in Celcius)")
    ax.grid(True)

fig.suptitle("Weekly Temperatures of Major Cities of the world")
fig.supxlabel("Days") # super x label, generalises x label of all graphs
fig.supylabel("Temperature (in Celcius)") # super y label, generalises y label of all graphs
fig.tight_layout()
```
