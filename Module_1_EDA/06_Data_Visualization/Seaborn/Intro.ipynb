# Seaborn
- Seaborn is a python library used for data visualisation and is based on matplotlib
- Seaborn is used when we need to draw/observe major details (high level)
- matplotlib is used when we need to draw/observe minute details (low level)
- seaborn and matplotlib are very very related
- for official documentation go to https://seaborn.pydata.org/
- seaborn benifits:-

    1. faster
    2. has in-built datasets
    3. has a lot of high level features
    4. has great support for pandas dataframe
    5. complex graphs like boxplots, heatmaps etc can be drawn pretty easily



```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# generally first we choose/set a theme
# there are 5 preset themes in seaborn, 1. dark grid(default), 2. white grid, 3. dark, 4. white, 5. ticks

sns.set_theme() 

# we are using a sns built in data set

sns.get_dataset_names() # function to see all the names of the dataset, we are choosing tips data
tips = sns.load_dataset("tips")
print(type(tips)) # it is actually a dataframe, so we can use all the dataframe knowledge and functions 

tips.head()
```
