# Categorical plots
### We analyse our data based on categories, for example:- (these are the types of relational plots)

1. bar plots
2. box plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# bar plot
# analysing amount of tip per day given by male and female

sns.set_theme()

tips = sns.load_dataset("tips")
tips.head()

sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# box plots
# analysing amount of tip per day given by male and female

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```
