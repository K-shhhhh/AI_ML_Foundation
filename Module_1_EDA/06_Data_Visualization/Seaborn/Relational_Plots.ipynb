# Relational Plots
### shows relationship between data points/axes, for example:- (these are the types of relational plots)

1. scatter plot
2. line plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid") 

tips = sns.load_dataset("tips")

# we plot the data for the amount of bill and how is it related to the amount of tip given

sns.relplot( # command to draw a relational plot, default kind is scatter plots
    data=tips, # telling which data do we need to pass 
    x = "total_bill", # here is where dataframe integration works, rather than writing tips["x"] we can simply write "x"
    y = "tip",
    # we can see that labels are there by default, automatically
    hue="time", # differentiates between data points for lunch/dinner using color, dont need to use legend as well
    style="time", # differentiates between data points for lunch/dinner using style, dont need to use legend as well
    size = "size", # giving size of each point to it via the dataframe
)
```

```python
# we can also use scatterplot

sns.scatterplot( 
    data=tips, 
    x = "total_bill",
    y = "tip"
)
```

```python
# line plot using seaborn
sns.set_theme(style="ticks") 

flights = sns.load_dataset("flights") # in built dataset 
flights.head()

# we plot the data for passengers per year in flights

sns.relplot(
    kind="line", # to plot a line plot we just do this 
    data=flights,
    x = "year",
    y = "passengers"
    # the shading around the main line is called confidence interval (ci)
    # it shows the probability of how much the data can lie in the confidence interval range
    # it is nothing but the value of mean +- ci
    # ci is by default taken as 95%
    
)
```

```python
# we can also use lineplot

sns.lineplot(
    data=flights,
    x = "year",
    y = "passengers"
    
)
```
