# Data Visualization

- Data Visualisation is for humans, a picture speaks more than a thousand words
  better for humans analyse some data as a picture/chart, rather than some file or table etc, those are for the computer

- Benifits of Data Visualization :-

    1. helps a lot in eda, find outlers
    2. helps in model evaluation
    3. better for communication of data to stake holders and higher authorities as they arent programmers like us
    4. helps in better business decisions or research decisions

### Types of graphs we plot:-

    1. line graph
    2. pie chart
    3. histogram
    4. scatter plots

- the main question always is which chart to choose for a particular data set to get the best visuals, therefore we shall remember
  these points when we have to choose which chart to use:-

   1. what is the actual problem? and what exactly do we need to show
   2. the data shall always have no/very less noise
   3. we have to do story telling with the graph, it should be understandable by anyone and everyone
   4. data should always and always have clarity, labels/ abbreviations/ units/ colors etc should be named and explained properly
