- we use seaborn and matplotlib together as a data scientist / AI Engineer
- we plot an individual graph using seaborn, and if we want to do any customizations then we use matplotlib
- look at an example:-

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# we create individual graphs using seaborn and then use modern matplotlib to edit the canvas (plt.subplots)

# Q. in the tips data, plot a graph of how much bill was paid each day by male and female respectively

sns.set_theme(style="white")

tips = sns.load_dataset("tips")
tips.head()

fig,ax = plt.subplots() # creating the figure and axes

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None, # removing the confidence interval
    ax=ax # optional, axis in which to draw the plot
)

# after creating the graph, for other minute details we use matplotlib

ax.set_title("Bill on each day by Men and Women") # using ax as its modern matplotlib
ax.set_xlabel("Days")
ax.set_ylabel("Amount")
ax.grid(True)
fig.tight_layout()
```

# The Cheat Sheet: Axes-level vs. Figure-level

### To use Seaborn professionally, you need to know which functions play nicely with Matplotlib's ax and which ones act as standalone managers.

1. Axes-Level Functions (The Polite Workers) : These draw specific, single charts. They always accept the ax=ax parameter so you can place them inside complex Matplotlib layouts.

   - sns.lineplot()

   - sns.scatterplot()

   - sns.histplot()

   - sns.boxplot()

   - sns.barplot()

2. Figure-Level Functions (The Control Freaks) : These are umbrella functions designed to create multi-plot grids automatically (using FacetGrid). They do not accept ax=ax. You must let them manage the whole figure.

- sns.relplot() (The umbrella for lineplot and scatterplot)

- sns.displot() (The umbrella for histplot and kdeplot)

- sns.catplot() (The umbrella for boxplot, barplot, violinplot)
