# Challenge 1: The Basics (Extraction & Looping)

- Target: http://quotes.toscrape.com/
- Scrape the front page.
- The Output: A pandas DataFrame with three columns: Quote, Author, and the First_Tag (just grab the first tag listed for each quote, not the whole list).

```python
import requests
from bs4 import BeautifulSoup 
import pandas as pd
```

```python
URL = "https://quotes.toscrape.com/"
response = requests.get(URL)

# if response.status_code == 200:
#     print(response.text)

with open("front_page.html","w") as f:
    f.write(response.text)
```

```python
with open("front_page.html","r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")

all_quotes_data = []

# Grab all the main quote containers
quotes_containers = soup.select("div.quote")

# Loop through the containers ONCE
for box in quotes_containers:
    
    # Look inside the current box for the specific elements
    text = box.select_one(".text").get_text()
    author = box.select_one(".author").get_text()
    
    # Handle the tag (with a safety check in case a quote has NO tags)
    tag_element = box.select_one(".tags a.tag")
    first_tag = box.select_one("a.tag").get_text() if tag_element else "None"    
    
    # Append as a row to our main list
    all_quotes_data.append([text, author, first_tag])

# Pop it directly into a dataframe
df = pd.DataFrame(all_quotes_data, columns=["Quotes", "Authors", "First_Tags"])

df
```
