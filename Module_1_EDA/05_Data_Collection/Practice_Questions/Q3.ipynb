# Challenge 3: The Boss Level (Pagination & Logic)
- Target: http://books.toscrape.com/
- Scrape a multi-page e-commerce layout.
- The Output: Write a loop that goes through the first 3 pages of the site and extracts the Title and Price of ONLY the books that have a "Five Star" rating (hint: you will need to look closely at the class names of the rating stars).

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
page_count = 1

while (page_count < 4):
    URL = f"https://books.toscrape.com/catalogue/page-{page_count}.html"
    response = requests.get(URL)

    if response.status_code == 200:
        with open(f"books_{page_count}.html","w") as f:
            f.write(response.text)
            print(f"page {page_count} downloaded")

    page_count = page_count + 1
```

```python
page_count = 1

five_star_books = []
while (page_count < 4):
    with open(f"books_{page_count}.html","r") as f:
        html_content = f.read()
    
    soup = BeautifulSoup(html_content, "lxml")
    
    all_books = soup.select("article.product_pod")
    
    for book in all_books:
        if book.select_one("p.star-rating.Three"): # when there is a space in the class name we use a.b but if we want to go inside
                                                   # an element to find one more we give a space in between like a. c
                
                title = book.select_one("h3 a")["title"]
                
                price = book.select_one("p.price_color").get_text(strip=True)
                
                five_star_books.append([title, price])

    page_count = page_count + 1

df = pd.DataFrame(five_star_books, columns=["Title" , "Price"])
```
