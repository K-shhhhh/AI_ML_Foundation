# Challenge 2: Tabular Data (Navigating Siblings)
- Target: https://www.scrapethissite.com/pages/forms/ (Hockey Teams)
- Scrape the table data for a specific team.
- The Output: Write a script where you input a team name (e.g., "Boston Bruins"), and it scrapes the page to return their "Year", "Wins", and "Losses" for the visible rows.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
URL = "https://www.scrapethissite.com/pages/forms/"
response = requests.get(URL)

# print(response.text)

with open("Hockey_teams.html","w") as f:
    f.write(response.text)
    print("file made successfully")
```

```python
team = input("Enter your favorite team: ")
print(f"Your favorite team is: {team}")
```

```python
with open("Hockey_teams.html","r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")

all_teams = []
all_wins = []
all_losses = []

table = soup.select("table.table .team")

for t in table:
    all_teams.append(t.select_one("td.name").get_text(strip=True))
    all_wins.append(t.select_one("td.wins").get_text(strip=True))
    all_losses.append(t.select_one("td.losses").get_text(strip=True))

df = pd.DataFrame({
    "Teams" : all_teams,
    "Wins" : all_wins,
    "Losses" : all_losses
})

df_team = df[df["Teams"] == team]

print(df_team)
```
