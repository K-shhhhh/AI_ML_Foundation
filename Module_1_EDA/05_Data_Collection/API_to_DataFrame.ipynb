```python
import requests
import pandas as pd
```

```python
# We will use requests to work with api
# We send a request to the API URL and the response is our data
# There are different types of requests like GET, POST etc but we generally use GET request

URL = "https://stephen-king-api.onrender.com/api/books"
response = requests.get(URL) 
```

```python
response # status code 200 means everything is working alright
```

```python
json_data = response.json() # stores the data in json format
```

```python
df = pd.json_normalize(json_data["data"]) # normalises the json_data, all of it was stored in "data" column
df = df[["id","Title","Year","ISBN"]]
```
