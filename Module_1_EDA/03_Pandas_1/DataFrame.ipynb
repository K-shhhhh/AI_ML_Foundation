```python
import pandas as pd
import numpy as np
```

```python
# 1. Creating dataframe with dict
info = {
    "Name" : ["Krish","Hiyu","Vero"],
    "Marks" : [91.4,92.4,91.4]
}

df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
```

```python
# 2. creating dataframe with lists of lists
df = pd.DataFrame([["Krish",91.4],["Hiyu",92.4],["Vero",91.4]], columns = ["Name","Marks"])
print(df)
print(df.index)
print(df.columns)
```

```python
# 3. creating dataframe from numpy array

np_array = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_array,columns = ["A","B","C"])
print(df)


```

```python
# working with csv data and pandas

df = pd.read_csv("employee_data.csv") # for excel use read_xlsx and so on

df.head(2) # .head() = prints first 5 rows only, but we can customize
df.tail(3) # .tail() = prints last 5 rows, but we can customize
df.sample(4) # .sample() = any random value, but we can customize
df.info() # .info() = a summary of our data
df.shape # .shape = gives the shape of our data as a tuple (rows,columns)
df.describe() # .describe() = gives us some features/information of the data for all the numerical columns 
df.columns # prints the indexes
df.nunique() # prints how many different types of values in each column
```

```python
# working with json data and pandas

df = pd.read_json("employee_data.json")
print(df, type(df))
```
