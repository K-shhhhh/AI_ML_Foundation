```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

## Handling missing values

```python
# there are multiple ways to handle missing values

df.isnull() # 1. gives true wherever there is a missing value 
df.isnull().sum() # 2. we can chain the .sum() to know how many missing values in each column
df.dropna() # 3. drops all the rows which contain a null value
df.dropna(axis=1) # 4. drops all the cols which contain a null value
df.fillna(0) # 5. fills all the missing values with a value of our choosing, in this case its 0
df.ffill() # 6. fills the previous row data if there is a missing value (top to bottom)
df.bfill() # 7. fills the next row data if there is a missing value (bottom to top)
```

## Handling duplicates

```python
df.duplicated() # 1. tells via true/false which row is duplicated
df["country"].duplicated() # we can check that for a particular column or combos of columns as well
df.drop_duplicates() # 2. drops the duplicated rows/ any particular selected cols or combos of cols

# to implement drop_duplicate or any imp function to the original data, use inplace=True in the brackets, still NOT advices
```
