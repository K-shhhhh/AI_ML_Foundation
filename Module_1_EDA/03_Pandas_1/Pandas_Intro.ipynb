```python
import pandas as pd
```

```python
# Pandas is based on numpy and used for EDA (Experimental Data Analysis)
# Series : 1D Labelled Array
# Dataframe : 2D Labelled Array (table)

info = {
    "Name" : ["Krish","Hiyu","Vero"],
    "Marks" : [91.4,92.4,91.4]
}

df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
print("\n")

series = pd.Series([1,2,3,4,5])
print(series)
```

```python
# Index : a position, it is a way to access a particular element in an array or series [0,1,2 etc]
# Label : a customized index, It is a way to represent each row in a series/dataframes [any name we want]
# In a series, index and label are same as it is a 1D array
```

```python
series = pd.Series([22,33,44,55], index = ["11*2","11*3","11*4","11*5"]) # these are the new labels
print(series)

# accesing through index
print(series[0])

# accessing through label
print(series["11*2"])

# printing all labels
print(series.index)
```
