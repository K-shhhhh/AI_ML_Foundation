```python
df = pd.read_csv("globalAirQuality.csv")
df.tail()
```

## Selecting Data

```python
# Selecting rows

# 1. loc (label based)
# 2. iloc (index based)

df.loc[0] # printing the first row
df.loc[0:2] # slicing [inclusive]

df.iloc[0] # in this dataset, index value and labels are same so the output would be same
df.iloc[0:2] # sliciing [exlusive]
```

```python
# selecting cells (scalar)

# 1. use columns and rows method together

df.loc[0,"aqi"] # answer is 108

# selecting some sets of cells

df.loc[0:2,["city","aqi"]] # using loc we passed the labels (strings)
df.iloc[0:2,[2,11]] # using iloc we have to pass index number and if we pass strings it will show an error

# 2. at (using labels) and iat (using indexes)

df.at[0,"city"] # using labels
df.iat[0,2] # using indexes
```

## Filtering data

```python
# Here we need to get data based on some conditions

df[df["country"]=="US"] # shows data where country is US only
df[(df["country"]=="US") & (df["aqi"]>100)] # shows data where country is US and aqi is more than 100
df[(df["country"]=="US") & (df["aqi"]>100)][["city", "aqi"]] # shows the same data but only the 2 columns of city and aqi (chaining)
```

## Queries

```python
# rule - 1 : pass expression as a string
# rule - 2 : column based reference (case sensitive)
# rule - 3 : use logical / relational operators (/,>,<,+,- etc)
# rule - 4 : can use chaining
# rule - 5 : can use normal python variables using @
# rule - 6 : can use multiple conditions/comparisions

aqi_limit = 100
df.query("aqi>@aqi_limit & country == 'US'")[["city","aqi"]]
```
