### Task : Clean the raw data, and write it into a new file 

```python
import pandas as pd
```

```python
# clean data = delete duplicate, sort, handle null values, preserve index values as well

df = pd.read_csv("raw_data.csv")
df2 = df.copy()
df2 = df2.drop_duplicates()
df2 = df2.fillna(0)
df2 = df2.sort_values("income")
df2 = df2.reset_index(drop=True)
```

```python
df2.to_csv("sorted_data.csv")
```
