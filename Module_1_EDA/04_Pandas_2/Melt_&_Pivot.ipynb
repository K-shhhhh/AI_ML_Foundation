```python
import pandas as pd
df = pd.DataFrame({
    "country" : ["USA","USA","India","India"],
    "year" : [2020,2022,2021,2020],
    "sales" : [100,120,90,110],
    "profit" : [20,25,18,22]
})

df
```

### Melt = Wide to Long Format

```python
melted_data = df.melt(
                id_vars=["country","year"], # these are columns we want to keep
                value_vars=["sales","profit"], # these are values we want to pivot, as in merge 
                var_name = "metrics", # this is the new column where the pivoted column names would be registered as data points
                value_name = "value" # this is the new column where all the data of the pivoted columns will be transferred into
)

melted_data
```

### Pivot = Long to Wide Format

```python
pivoted_data = melted_data.pivot(
    index = ["country","year"], # The columns we want to keep, unique values
    columns = "metrics", # The extra columns we want to have
    values = "value"  # the corresponding data in the extra columns we will create
)

pivoted_data
```
