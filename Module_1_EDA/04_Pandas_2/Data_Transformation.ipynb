### Feature engineering : when we transform our data to form new columns so that we improve the analystics and features of our data

```python
# Q. make a new column as tax, if income is greater than 50k then 20% tax if not then 10%
df2 = df.copy()
df2["Gender"] = ["Female","Male","Male","Male","Female","Male","Female","Female","Male","Female"] # just makign a new column
df2["tax"] = df2["Salary"].apply(lambda x : "20%" if x >= 50000 else "10%") # 1. apply function : use another function in it like lambda
df2
```

```python
# Q. make female to F and Male to M using a function

sex_map = {
    "Male" : "M",
    "Female" : "F",
    "Unknown" : "U"
}

df2["Gender"] = df2["Gender"].map(sex_map) # 2. .map is used to convert a value into another in a column
df2
```

```python
# Q. Make a new column as new salary where every employee has a promotion by 50%

df2.assign(new_salary = df2["Salary"] * 1.5) # 3. .assign makes a new column on the basis of an old column
```

```python
# Q. Replace the country name "USA" to "US"
df2["Department"] = df2["Department"].replace("IT","ICT") # 4. .replace replaces all the entries to a customized new one in a column
df2
```

```python
# Q. Replace the name of the column Gender to Sex and the row label 1 to 1st
df2 = df2.rename(columns={"Gender" : "Sex"}) # 5. .rename helps us change the column name of the row label
df2 = df2.rename(index={1 : "1st"})
df2
```

```python
# Q. Sort data on the basis of income and age but again sort in on the basis of index as thats most imp

df2.sort_values("Salary",ascending=False)
sorted_df = df2.sort_values(["Salary","Age"]) # 6. .sort_values sorts values in a col, can pass more parameters are lists
sorted_df.sort_index() # 7. .sort_index sorts on the basis of index values

# Q. In this same data,reset it to original while preserving the sorting and then drop the indexes

sorted_df.reset_index()
sorted_df.reset_index(drop=True)
```

```python
# Q. give ranks to each employee based on income

sorted_df["ranking"] = sorted_df["Salary"].rank(ascending=False) 
sorted_df
```
