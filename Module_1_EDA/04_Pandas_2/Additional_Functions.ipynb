### Aggregation and Group By functions

```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
df
```

```python
# everytime with an groupby function, we should always have another math function with it
df.groupby("country")["income"].max() 
df.groupby("age")["income"].min()
df.groupby("gender")["income"].mean() 
```

```python
# agg/aggregate function is the best, does everything in one only
# we can rename cols, calculate functions in a single line etc all in this function
df.groupby("country")["income"].agg(["mean","max","min"]) 
df.groupby("country")["income"].aggregate(["mean","max","min"]) 
df.groupby("country").agg(
    max_salary=("income", "max"), 
    mean_age=("age", "mean")
)
```
