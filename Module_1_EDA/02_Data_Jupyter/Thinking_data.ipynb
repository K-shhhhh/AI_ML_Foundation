# Assigment-1 : clean and structure data

```python
import json
```

```python
# load data

def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f) 
        
    return data
```

```python
data = load_data("store_data.json")
print(data, type(data))
```

#### As we can see the data is very inconsistent

```python
# cleaning data

def clean_data(data):
    text_to_num  = {
        "one" : 1,
        "two" : 2,
        "three" : 3,
        "four" : 4,
        "five" : 5
    }
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # clean rating
        raw_rating = str(user["rating"]).strip().lower() # to clear white space, for lowercase,
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating] # replace the string to the key in text_to_map dictionary

        user["rating"] = raw_rating

        # handle missing values
        raw_age = user.get("age") # get functions will give us any value for the key "age" and if no key is there then gives None
        if raw_age == None:
            user["age"] = None

        # handling duplicate entries
        if user["name"].strip() in unique_users:
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user) 

    return cleaned_data
```

```python
data = clean_data(data)
```

# Assignment-2 : Getting meaningful insights from data

### 1. find avg rating
### 2. % of users gave rating less than 3 (poor)

```python
def get_insights(data):

    # avg_rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])

    print(f"Average rating : {total_rating/len(data)}")

    # poor ratings
    poor_ratings = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_ratings += 1

    print(f"poor rating = {(poor_ratings/len(data)) * 100}%")
    
```

```python
get_insights(data)
```

# Assignment-3 : Product reccomendation feature
### 1. If user rating is >= 4, recommend same brand
### 2. If user rating is <= 4, recommend different brand

```python
def get_recommendation(data):
    recommendations = []

    for user in data:
        curr_recommendation = {}
        curr_recommendation["name"] = user["name"]
        if float(user["rating"]) >= 4:
            curr_recommendation["Brand"] = "Apple"
        else:
            curr_recommendation["Brand"] = "Samsung"

        recommendations.append(curr_recommendation)
    return recommendations       
```

```python
get_recommendation(data)
```
