## Scratch Implementation of Logistic Regression (Classification)

```python
# LGR converts LNR values to probabilities i.e. between 0 to 1 using the sigmoid function
# we will use the exact same steps as LNR but with a few tweaks

# If a function or variable belongs inside a class, it almost always needs self (example :- self.probabilites)
# must use self for variables we want the object to remember permanently (example :- bias, weights etc)
# not use self for temporary variables that only exist inside a single loop or function (example :- m, n, y_pred)
```

```python
import numpy as np
```

```python
class LogisticRegression():

    # first we will create the constructor

    def __init__(self, learning_rate = 0.1, n_iter = 1000):
        self.bias = None
        self.weights = None
        self.n_iter = n_iter
        self.lr = learning_rate


    # create sigmoid function

    def _sigmoid(self, z) : 
        return(1 / (1 + np.exp(-z))) # np.exp gives us e (eulers number) and we give its power as -z

    
    # create fit function

    def fit(self, X, y):

        # get shape of X
        
        m, n = X.shape
        
        # step 1 : randomly initialise

        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.n_iter) :
            
            # step 2 : make predictions
            z = self.bias + (X @ self.weights)
            y_pred = self._sigmoid(z) # these arent predictions like 0 or 1 but are probabilities like 0.1, 0.7 etc
            
            # step 3 : calculate gradient
            db = (1/m) * np.sum(y_pred - y) 
            dw = (1/m) * (X.T @ (y_pred - y)) # we dont do np.sum here as X is a matrix, (y_pred - y) will have dimensions of (m,1)
                                              # and X will have the dimensions (m, n) so make them compatible with dot product we 
                                              # do X.T so that (n , m) * (m , 1) = (n, 1)
    
            # step 4 : convergence theorum
            self.bias = self.bias - (self.lr * db)
            self.weights = self.weights - (self.lr * dw)


    # create a function to get probabilities

    def get_probabilities(self, X): # this will be pure probabilities like 0.1, 0.7 etc
        
        z = self.bias + (X @ self.weights)
        return (self._sigmoid(z))
    
    # create predict function 

    def predict(self, X, threshold_value = 0.5): # this will be actual predictions for an input like 0 and 1 only

        prob = self.get_probabilities(X) # we get the probabilities from the function
        return np.where(prob >= 0.5, 1, 0) # this is a numpy function " where " --> (condition, if true value, else value)
```

```python
X = np.array([[1,2], [2,3], [3,4], [4,5]])
y = np.array([0, 0, 1, 1])

lgr_model = LogisticRegression()
lgr_model.fit(X, y)
y_pred = lgr_model.predict(X)

print(y_pred)
print(lgr_model.get_probabilities(X))
```
