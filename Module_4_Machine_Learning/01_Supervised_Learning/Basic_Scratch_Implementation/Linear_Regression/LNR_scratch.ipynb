## Linear regression via Gradient Descent

```python
# in scratch implementation, we will know why vectors, matrices etc are so very imp
# @ = matrix multiplication, * = element-wise multiplication, (.) = access matrix built in methods
```

```python
import numpy as np

class LinearRegression:

    # first we will define the constructor
    # theta 0 : bias
    # theta 1 / theta 2 .... : coefficent or weight
    # every model will have different bias and weight
    # we need to initialise is learning rate (alpha), and we randomly choose it, like here its 0.01
    # one last thing we need to initialise is max iterations for GD 

    def __init__(self, learning_rate = 0.01, n_iter=1000): 
        self.bias = None 
        self.weights = None 
        self.lr = learning_rate
        self.n_iter = n_iter

    # create the fit function
    
    def fit(self, X, y): # we can call it X_train or y_train as well
        
        m, n = X.shape # this returns number of samples (m) and number of features (n) like rows and cols

        # STEP 1 : randomly initialise
        self.bias = 0 # initial bias with 0
        self.weights = np.zeros(n) # we dont know how many features are there so we cant put a scalar 0, so we make an array of 0's

         # this is the gradient descent loop to converge towards the minima
        for i in range(self.n_iter) : 
        
            # STEP 2 : make predictions
            y_pred = self.bias + (X @ self.weights) # we want to calculate the dot product of 2 matrix X and vector weights  
                                                    # i.e. every x in X should be multiplied by the corresponding w in weights
    
            # STEP 3 : calculate gradient 
            # db : derivative with respect to bias (theta 0 )
            # dw : derivative with respect to weights (theta 1)
            db = (1/m) * np.sum(y_pred - y)   # np.sum is just summation just like we use in formula of db
            dw = (1/m) * (X.T @ (y_pred - y)) # We want to multiply X with (y_pred - y) but as both are matrices we do 
                                              # dot product. (y_pred - y) will have dimensions of (m,1) and X will have the 
                                              # dimensions (m, n) so make them compatible with dot product we do X.T so that 
                                              # (n , m) * (m , 1)=(n, 1).
    
            # STEP 4 : convergence theorum
            self.bias = self.bias - (self.lr * db)
            self.weights = self.weights - (self.lr * dw)
        

    # create the predict function
    
    def predict(self, X):

        # these are the predictions that we make after finding the correct value of bias and weights from the fit() function
        y_pred = self.bias + (X @ self.weights)
        return y_pred
        
```

```python
# test the model we made

X = np.array([[1], [2], [3] , [4] , [5]]) # 2 brackets as X is supposed to be a 2d array (rows and cols, m and n)
y = np.array([2,4,6,8,10]) # only 1 bracket as y is suppose to be just a 1d array (a single col, n)

lnr_model = LinearRegression()
lnr_model.fit(X, y)

y_pred = lnr_model.predict(X)

print(y_pred)

print(lnr_model.bias) # should be 0, y = 2x + 0 is our best fit line
print(lnr_model.weights) # should be 2
```

## Linear Regression via OLS

```python
# we use OLS to implement Linear Regression

class LinearRegressionOLS():

    # here also we need to build constructor but no need of learning rate (alpha) or number of iterations
    
    def __init__(self):
        self.bias = None
        self.weights = None

    # here is the major change
    def fit(self, X, y):
        m , n = X.shape

        # step 1 
        X_updated = np.c_[np.ones((m,1)), X] # we want the add a new column in X at 0th index and the values in it should be 1
                                           # np.c_ is a function that take 2 matrices A, B and just merges them side by side as AB

        # step 2
        theta = (np.linalg.inv(X_updated.T @ X_updated)) @ (X_updated.T) @ y # this is just the normal equation we studied

        # step 3
        self.bias = theta[0]
        self.weights = theta[1:] # this means everything after that index 1
        


    # this will exactly be the same 
    def predict(self, X):
        y_pred = self.bias + (X @ self.weights)
        return y_pred
```

```python
# test the model we made

X = np.array([[1], [2], [3] , [4] , [5]]) # 2 brackets as X is supposed to be a 2d array (rows and cols)
y = np.array([2,4,6,8,10]) # only 1 bracket as y is suppose to be just a 1d array (a single col)

lnr_ols_model = LinearRegressionOLS()
lnr_ols_model.fit(X, y)

y_pred = lnr_ols_model.predict(X)

print(y_pred)

print(lnr_ols_model.bias) # should be 0, y = 2x + 0 is our best fit line
print(lnr_ols_model.weights) # should be 2
```
