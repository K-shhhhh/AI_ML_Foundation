- the math and some of the logic behind all ML algo are already instilled in these libraries to make our coding very easy
- for documentation and help go to : https://scikit-learn.org/stable/
- benifits of using skicit-learn :-

  1. well-documented
  2. easy to learn
  3. works well with numpy and pandas
  4. used by professionals and beginners
