# Machine Learning

- ML is a part of computer science where we teach the computer to learn from data, find patters and make decisions or predictions

- Tranditional Computer science Flow : INPUT ---> LOGIC (PROGRAM) ----> OUTPUT

- ML Flow : INPUT AND OUTPUT ----> ML MODEL (PROGRAM) -----> LOGIC OR HYPOTHESIS
- after we get this logic, we feed in new input to get new outputs

# Types of ML : -

1. Supervised Learning : Labelled data is used to train the model
    - label data : for every input value an output value is available
    - There are basically 2 problems we solve in supervised learning :-
---
    A. Classification : predict the categories/classes (spam/not spam)
    B. Regression : predict continuous value (aqi/stock price)
 

2. Unsupervised Learning : Unlabelled data is used to train the model
   - unlabelled data : for input value there arent output values
   - The types of problem we solve in unsupervised learning are:-
---
    A. Clustering : cluster related data together and identify patterns (amazon : phone+ charger)

3. Reinforced Learning : used when we have a goal oriented job, we generally mix it with supervise and unsupervised learning
    - We work with rewards and penalties, give rewards when the machine does something right and penalty when it does something wrong
