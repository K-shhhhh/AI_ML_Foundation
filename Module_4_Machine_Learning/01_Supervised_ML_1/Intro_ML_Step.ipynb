# Supervised Machine learning
- in this one we train the algorithm on labeled datasets, meaning each input has a known correct output to learn patterns and predict outcomes for new data
- we represent any information about something given as inputs (X or x)
- we represent the correct answer based on the input data is called output(Y or y)
- generally the data is given in a 2d table (pd.DataFrame)
- each row is a single datapoint as in the table is about emails, then each row is information (inputs) of a single email
- each column is called features, as they are the characterstics of the data 
    - independent features dont depend on any other values i.e. they are the inputs (X)
    - dependent features are dependent on some other inputs .i.e. they are the outputs (Y)
---

### steps to build a machine learning model :-
    1. Data Cleaning
    2. Split the Data into train/test/validation set
    3. pre processing and encode data (numbers)
    4. feature engineering
    5. select which model to use/apply
    6. train on training set
    7. fine tune (overfitting or underfitting) using validation set
    8. test on test data for final output 
    9. make fixes and deploy

  

1. Regression : output is a continous numerical value

2. Classification : output is a category or class
