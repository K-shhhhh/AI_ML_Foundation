# Machine Learning

- ML is a part of computer science where we teach the computer to learn from data, find patters and make decisions or predictions

- Tranditional Computer science Flow : INPUT ---> LOGIC (PROGRAM) ----> OUTPUT

- ML Flow : INPUT AND OUTPUT ----> ML MODEL (PROGRAM) -----> LOGIC OR HYPOTHESIS
- after we get this logic, we feed in new input to get new outputs

# Types of ML : -

1. Supervised Learning : Labelled data is used to train the model
    - label data : for every input value an output value is available
    - There are basically 2 problems we solve in supervised learning :-
        - Classification : predict the categories/classes (spam/not spam)
        - Regression : predict continuous value (aqi/stock price)
 

2. Unsupervised Learning : Unlabelled data is used to train the model
   - unlabelled data : for input value there arent output values
   - The types of problem we solve in unsupervised learning are:-
        - Clustering : cluster related data together and identify patterns (amazon : phone+ charger) etc

3. Reinforced Learning : used when we have a goal oriented job, we generally mix it with supervise and unsupervised learning
    - We work with rewards and penalties, give rewards when the machine does something right and penalty when it does something wrong

# Supervised Machine learning
- in this one we train the algorithm on labeled datasets, meaning each input has a known correct output to learn patterns and predict outcomes for new data
- we represent any information about something given as inputs (X or x)
- we represent the correct answer based on the input data is called output(Y or y)
- generally the data is given in a 2d table (pd.DataFrame)
- each row is a single datapoint as in the table is about emails, then each row is information (inputs) of a single email
- each column is called features, as they are the characterstics of the data 
    - independent features dont depend on any other values i.e. they are the inputs (X)
    - dependent features are dependent on some other inputs .i.e. they are the outputs (Y)

### steps to build a machine learning model :-
    1. Data Cleaning
    2. Split the Data into train/test/validation set
    3. pre processing and encode data (numbers)
    4. feature engineering
    5. select which model to use/apply
    6. train on training set
    7. fine tune (overfitting or underfitting) using validation set
    8. test on test data for final output 
    9. make fixes and deploy
